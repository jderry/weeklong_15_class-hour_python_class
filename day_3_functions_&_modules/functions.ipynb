```python
# get numerical integration by passing function to integrate as argument into integration function
from numpy import linspace, sum as s
def getNumIntegral(f, end_x, start_x=0, delta_x=1e-8):
    x = linspace(start_x, end_x, int((end_x - start_x) / delta_x))
    return s(f(x) * delta_x)
```

```python
from scipy.integrate import quad
```

```python
# lambda returns x**2 given x; integrate from x = 0 to 5
getNumIntegral(lambda x: x**2, 5)
```

```python
# same lambda
quad(lambda x: x**2, 0, 5)
```

```python
# get numerical derivative
def getNumDerivative(f, x, h=1e-11):
    return (f(x + h) - f(x)) / h
```

```python
from scipy.differentiate import derivative
```

```python
getNumDerivative(lambda x: x**2, 5)
```

```python
derivative(lambda x: x**2, 5)
```
