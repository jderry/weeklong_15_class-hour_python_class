```python
import pandas as pd
```

```python
starCatalog = pd.read_csv('../dataset/hygdata_v3.csv.zip',\
                          index_col='id', compression='zip')
```

```python
starCatalog.fillna({'con': ''}, inplace=True)
```

```python
# if a star's record shows it is in constellation Ursa Major AND that it has a proper name...
mask = (starCatalog['con'].str.contains('UMa'))\
     & (starCatalog['proper'].notna())
```

```python
# then display it in a mini-table with columns 'proper', 'con', 'x', 'y', 'z' from the starCatalog
starCatalog[mask][['proper', 'con', 'x', 'y', 'z']]
```
